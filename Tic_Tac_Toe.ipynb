```python
import random

def print_board(board):
    print()
    for i in range(3):
        print(" " + " | ".join(board[i * 3:i * 3 + 3]))
        if i < 2:
            print("---+---+---")
    print()


def check_winner(board, symbol):
    winning_combinations = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    for combination in winning_combinations:
        if all(board[i] == symbol for i in combination):
            return True

    return False


def get_move(board, player_name):
    while True:
        try:
            move = int(input(f"{player_name}, enter your move (1-9): ")) - 1

            if move < 0 or move > 8:
                print("Please enter a number between 1 and 9.")
            elif board[move] != " ":
                print("That position is already occupied. Try again.")
            else:
                return move

        except ValueError:
            print("Invalid input! Please enter a number.")


def tic_tac_toe():
    print("=" * 35)
    print("       WELCOME TO TIC-TAC-TOE")
    print("=" * 35)

    # Select game mode
    while True:
        print("\nChoose Game Mode:")
        print("1. One Player (Play against Computer)")
        print("2. Two Players (Play with a Friend)")

        choice = input("Enter your choice (1 or 2): ").strip()

        if choice in ("1", "2"):
            break

        print("Invalid choice! Please select 1 or 2.")

    # Get player names
    player1 = input("\nEnter Player 1 name: ").strip()
    if not player1:
        player1 = "Player 1"

    if choice == "1":
        player2 = "Computer"
        print(f"\n{player1} (X) vs {player2} (O)")
    else:
        player2 = input("Enter Player 2 name: ").strip()
        if not player2:
            player2 = "Player 2"

        print(f"\n{player1} (X) vs {player2} (O)")

    board = [" "] * 9
    current_player = 1

    print("\nBoard positions:")
    print_board([str(i + 1) for i in range(9)])

    # Main game loop
    while True:
        print_board([
            str(i + 1) if board[i] == " " else board[i]
            for i in range(9)
        ])

        if current_player == 1:
            move = get_move(board, player1)
            symbol = "X"
            board[move] = symbol
            current_name = player1

        else:
            symbol = "O"
            current_name = player2

            if choice == "1":
                print("Computer is thinking...")
                available_moves = [
                    i for i in range(9) if board[i] == " "
                ]
                move = random.choice(available_moves)
                board[move] = symbol
                print(f"Computer chose position {move + 1}.")

            else:
                move = get_move(board, player2)
                board[move] = symbol

        # Check for winner
        if check_winner(board, symbol):
            print_board(board)
            print(f"Congratulations, {current_name}! You win!")
            break

        # Check for draw
        if " " not in board:
            print_board(board)
            print("It's a draw! Well played, both players.")
            break

        # Switch turns
        current_player = 2 if current_player == 1 else 1

    print("\nThank you for playing Tic-Tac-Toe!")


# Start the game
if __name__ == "__main__":
    tic_tac_toe()
```
