```python
import random

Q = [[random.randint(2, 25) for _ in range(2)] for _ in range(4)]

awt = lambda q: sum(sum(c[:i]) for c in q for i in range(len(c))) / sum(map(len, q))
print(f"Before: {awt(Q):.2f}m", Q)

for _ in range(30):
    W = [sum(c) for c in Q]
    moves = [(i, min([n for n in (i-1, i+1) if 0 <= n < 4], key=lambda x: W[x]))
             for i in range(4) if len(Q[i]) > 1]
    done = set()
    for s, d in moves:
        if W[s] - W[d] > 5 and not ({s, d} & done):
            Q[d].append(Q[s].pop()); done |= {s, d}

print(f"After:  {awt(Q):.2f}m", Q)
```
