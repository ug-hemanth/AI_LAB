```python
# 8-Puzzle using Depth-First Search (DFS)

start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def display(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def get_neighbors(state):
    neighbors = []
    zero = state.index(0)

    moves = {
        "Up": -3,
        "Down": 3,
        "Left": -1,
        "Right": 1
    }

    for move, change in moves.items():
        new_pos = zero + change

        if not (0 <= new_pos < 9):
            continue

        if move == "Left" and zero % 3 == 0:
            continue

        if move == "Right" and zero % 3 == 2:
            continue

        new_state = list(state)
        new_state[zero], new_state[new_pos] = (
            new_state[new_pos], new_state[zero]
        )

        neighbors.append((tuple(new_state), move))

    return neighbors


def dfs(start, goal, max_depth=30):
    stack = [(start, [])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            return path

        if state in visited:
            continue

        visited.add(state)

        if len(path) >= max_depth:
            continue

        for next_state, move in get_neighbors(state):
            if next_state not in visited:
                stack.append((next_state, path + [move]))

    return None


print("Initial State:")
display(start)

print("Goal State:")
display(goal)

solution = dfs(start, goal)

if solution is not None:
    print("Solution using DFS:", solution)
    print("Number of moves:", len(solution))
else:
    print("No solution found within the depth limit.")
```
