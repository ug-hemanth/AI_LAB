```python
# Simple Vacuum Cleaner Agent

# Environment: Two rooms A and B
rooms = {
    'A': 'Dirty',
    'B': 'Dirty'
}

# Starting position of the vacuum cleaner
current_location = 'A'

print("Vacuum Cleaner Agent")
print("--------------------")

while True:
    print("\nCurrent Location:", current_location)
    print("Room Status:", rooms)

    # Check whether the current room is dirty
    if rooms[current_location] == 'Dirty':
        print("Room", current_location, "is dirty.")
        print("Cleaning room", current_location, "...")
        rooms[current_location] = 'Clean'
        print("Room", current_location, "is now clean.")

    else:
        print("Room", current_location, "is already clean.")

    # Move to the other room if it has not been cleaned
    if current_location == 'A':
        current_location = 'B'
    elif current_location == 'B':
        if rooms['A'] == 'Clean' and rooms['B'] == 'Clean':
            break
        current_location = 'A'

    # Stop when both rooms are clean
    if rooms['A'] == 'Clean' and rooms['B'] == 'Clean':
        break

print("\nFinal Room Status:", rooms)
print("Both rooms are clean!")
print("Vacuum Cleaner Agent has completed its task.")
```
