```python
import random
import math

N = 8

# Calculate the number of conflicts between queens
def heuristic(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            # Check diagonal conflicts
            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Display the chessboard
def display_board(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()
    print()


# Simulated Annealing algorithm
def simulated_annealing():
    # Randomly place one queen in each column
    board = [random.randint(0, N - 1) for _ in range(N)]

    temperature = 100.0
    cooling_rate = 0.995
    min_temperature = 0.01
    step = 0

    print("Initial Board:")
    display_board(board)
    print("Initial Conflicts:", heuristic(board))

    while temperature > min_temperature:
        current_conflicts = heuristic(board)

        # Stop if a solution is found
        if current_conflicts == 0:
            print("Solution Found!")
            return board

        # Create a neighboring board by moving one queen
        new_board = board.copy()
        col = random.randint(0, N - 1)
        new_row = random.randint(0, N - 1)

        while new_row == new_board[col]:
            new_row = random.randint(0, N - 1)

        new_board[col] = new_row

        new_conflicts = heuristic(new_board)
        difference = new_conflicts - current_conflicts

        # Accept better moves or sometimes accept worse moves
        if difference < 0 or random.random() < math.exp(-difference / temperature):
            board = new_board
            step += 1

            print(f"Step {step}:")
            display_board(board)
            print("Conflicts:", heuristic(board))
            print("Temperature:", round(temperature, 2))
            print()

        # Gradually reduce the temperature
        temperature *= cooling_rate

    if heuristic(board) == 0:
        print("Solution Found!")
        return board

    print("Cooling completed. Restarting with a new board...")
    return None


# Restart if the algorithm does not find a solution
solution = None

for attempt in range(1, 101):
    print(f"\n========== Attempt {attempt} ==========")
    solution = simulated_annealing()

    if solution is not None:
        break

if solution is not None:
    print("\nFinal Solution:")
    display_board(solution)
    print("Final Conflicts:", heuristic(solution))
else:
    print("No solution found after 100 attempts.")
```
