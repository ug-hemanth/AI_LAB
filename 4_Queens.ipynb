```python
import random

N = 4

# Calculate the number of attacking queen pairs
def heuristic(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j]:
                conflicts += 1

            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Display the chessboard
def display_board(board):
    for row in range(N):
        line = ""
        for col in range(N):
            if board[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)
    print()


# Generate neighboring states
def get_neighbors(board):
    neighbors = []

    for col in range(N):
        for row in range(N):
            if board[col] != row:
                new_board = board.copy()
                new_board[col] = row
                neighbors.append(new_board)

    return neighbors


# Hill Climbing algorithm
def hill_climbing():
    board = [random.randint(0, N - 1) for _ in range(N)]
    step = 0

    print("\nInitial Board:")
    display_board(board)
    print("Conflicts:", heuristic(board))

    while True:
        current_h = heuristic(board)

        if current_h == 0:
            print("Solution Found!")
            return board

        neighbors = get_neighbors(board)
        best_board = board
        best_h = current_h

        for neighbor in neighbors:
            h = heuristic(neighbor)

            if h < best_h:
                best_h = h
                best_board = neighbor

        # Stop if no better neighbor exists
        if best_h >= current_h:
            print("Local optimum reached. Restarting...")
            return None

        board = best_board
        step += 1

        print(f"Step {step}:")
        display_board(board)
        print("Conflicts:", heuristic(board))


# Run hill climbing with random restarts
solution = None

for attempt in range(1, 101):
    print(f"\n========== Attempt {attempt} ==========")
    solution = hill_climbing()

    if solution is not None:
        break

if solution is not None:
    print("\nFinal Solution:")
    display_board(solution)
    print("Final Conflicts:", heuristic(solution))
else:
    print("No solution found after 100 attempts.")
```
